# AEMET Data Extraction (Weather data)

This notebook serves as a proof-of-concept to extract weather predictions from the AEMET Open Data API.

## Weather prediction by city

The API allows to retrieve weather predictions by city hourly and daily through the following endpoints:
- /api/prediccion/especifica/municipio/diaria/{municipio}
- /api/prediccion/especifica/municipio/horaria/{municipio}

More info can be found [here](https://opendata.aemet.es/dist/index.html?#/predicciones-especificas)

```python
from dotenv import load_dotenv
import os

load_dotenv()

AEMET_API_KEY = os.getenv("AEMET_API_KEY")

city_code = "08019" # Barcelona

base_url = f"https://opendata.aemet.es/opendata"
```

The city code can be retrieved from [this](https://www.ine.es/daco/daco42/codmun/diccionario24.xlsx) table and is formed by combining `CPRO` and `CMUN` columns.

```python
# Based on provided example: https://opendata.aemet.es/centrodedescargas/ejemProgramas?

import requests

url = f"{base_url}/api/prediccion/especifica/municipio/horaria/{city_code}"
querystring = {"api_key": AEMET_API_KEY}
headers = {
    'cache-control': "no-cache"
}

response = requests.request("GET", url, headers=headers, params=querystring)

print(response.text)
```

```python
import json

response_json = json.loads(response.text) 

data = requests.request("GET", response_json.get("datos"), headers=headers, params=querystring)
metadata = requests.request("GET", response_json.get("metadatos"), headers=headers, params=querystring)
```
