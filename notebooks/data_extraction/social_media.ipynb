# Social Media data extraction

## BlueSky (Hot + Warm Path)

### Goals
- Real-Time analysis of incidences and road congestion.
- Real-Time sentiment analysis on posts
- Aggregation of posts metrics (number of topics, number of posts per topic, number of positive neutral and negative comments, etc.)

```python
from atproto import Client
import os

def query_by_topic(topic: str, max_results: int = 10) -> list(dict()):
    client = Client()

    user = os.getenv("BLUESKY_USERNAME")
    passwd = os.getenv("BLUESKY_PASSWORD")

    try:
        client.login(user, passwd)
        print("✅ Inicio de sesión exitoso.\n")
        
        print(f"🔍 Buscando posts con: '{topic}' (Orden: Más recientes primero)...\n")
        
        # Llamada al endpoint de búsqueda de posts
        response = client.app.bsky.feed.search_posts({
            'q': topic,
            'sort': 'latest', # 'latest' ordena de más nuevo a más viejo
            'limit': max_results
        })

        # 4. Iterar sobre los resultados e imprimir la información
        if not response.posts:
            print("No se encontraron resultados para esta búsqueda.")
            return None

        res = []
        for post in response.posts:
            res.append({
                'author': post.author.handle,
                'text': post.record.text,
                'created_at': post.record.created_at
            })
            
            # Limpiamos los saltos de línea para que se lea mejor en la consola
            #texto_limpio = texto.replace('\n', ' ')
        return res

    except Exception as e:
        print(f"❌ Ocurrió un error: {e}")
        return None
```

```python
from dotenv import load_dotenv

load_dotenv()

topics = [
    "transit bcn",
]

for t in topics:
    posts = query_by_topic(t)
    for p in posts:
        print(p)
```
