```python
import chromadb

client = chromadb.HttpClient(host="chromadb", port=8000)

# Llista col·leccions
print(client.list_collections())

# Accedeix a una col·lecció
coll = client.get_collection("twitter_transit_posts")
print(f"Total documents: {coll.count()}")

# Veure alguns documents
results = coll.peek(10)
print(results)
```
