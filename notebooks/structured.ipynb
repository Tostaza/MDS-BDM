```python
load_dotenv()
minio_client = boto3.client(
    "s3",
    aws_access_key_id=os.getenv("ACCESS_KEY_ID"),
    aws_secret_access_key=os.getenv("SECRET_ACCESS_KEY"),
    endpoint_url= "http://" + os.getenv("API_ENDPOINT")
)

new_bucket = "landing-zone"
try:
    minio_client.create_bucket(Bucket=new_bucket)
except (minio_client.exceptions.BucketAlreadyExists, minio_client.exceptions.BucketAlreadyOwnedByYou):
    print(f"Bucket '{new_bucket}' already exists")
```

```python
import requests
import pandas as pd

BEARER = "TU_BARER_TOKEN"

def x_get(url, params=None):
    headers = {"Authorization": f"Bearer {BEARER}"}
    r = requests.get(url, headers=headers, params=params, timeout=30)
    r.raise_for_status()
    return r.json()


username = "transit"
user = x_get(f"https://api.x.com/2/users/by/username/{username}")
user_id = user["data"]["id"]

# 2) Últims 20 tweets/posts
tl = x_get(
    f"https://api.x.com/2/users/{user_id}/tweets",
    params={
        "max_results": 1,
        "tweet.fields": "created_at,lang",
        # "exclude": "replies,retweets"  # descomenta si vols només posts originals
    },
)

df = pd.DataFrame(tl.get("data", []))

# 3) Filtre Barcelona (ajusta paraules clau segons necessitis)
pattern = r"(?:\bBarcelona\b|Nus Trinitat|Ronda(?:\s+Litoral|\s+de\s+Dalt)|B-10|B-20)"
df_bcn = df[df["text"].str.contains(pattern, case=False, na=False, regex=True)].copy()

df_bcn[["created_at", "text"]]
```
